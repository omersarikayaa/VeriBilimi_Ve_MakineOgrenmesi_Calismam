WORKING WITH MISSING DATA

```python
weather_na_df = pd.read_excel("6-weatherna.xlsx")  #diğer eksik datalı excel dosyasını içeri çektik
```

```python
weather_na_df                                      #gördüğün gibi eksik kısımlar var Nan olan
```

```python
weather_na_df.isna()   #bu sefer boş olanlar oldugundan truelar var
```

```python
weather_na_df.describe()   #nan olmasını bunu yapamayacağımız anlamına gelmez elimizdekilerle bunu yapabiliriz
```

```python
weather_na_df["Paris"].count()   #spesifik mesela countu sadece paris veya amsterdam icin calıstırabiliriz mesela pariste 
                                 #6 data var amsterdamda 8 data var gibi 
```

```python
weather_na_df["Amsterdam"].count()
```

```python
weather_na_df["Amsterdam"].isna()   #aynı şekilde sadece paris icin isna çalıştırarak missing dataları görebiliriz
```

```python
weather_na_df.dropna()  #9 datadan 4 e düştü nan olan satırları sildi direkt 1 inde bile varsa
```

```python
weather_na_df.dropna(axis=1) #sütunun icinde boş varsa sil dedik satır kalmadı axis=1 sütuna bak demek
```

```python
weather_na_df
```

```python
weather_na_df.drop("Paris",axis=1)  #paris sütununu düşürdüm mesela sadece
```

```python
weather_na_df.fillna(20)      #olmayan değerleri Nan olanlar 20 ile doldur dedım
```

```python
weather_na_df.mean()    #sütunların ortalamasını aldık burada
```

```python
weather_na_df.median()  #burada da sütunların medianını aldı 
```

```python
weather_na_df.fillna(weather_na_df.mean())     #boş olan yerleri mean ile doldurdur dedik AMA
                                               #DİKKAT hepsine aynı ortalamayı koymuyor kendi sütunlarının ortalamasını koyuyor
                                               #aşağıdaki yazdıklarımı da koyabilirim mesela .max .min mantıklı degılde yine de yapılabilir bil
```

```python
weather_na_df.max()
```

```python
weather_na_df.min()
```

```python
df = pd.read_csv("6-employee.csv")
```

```python
df      #veri çoğaldı bir şeyleri seçebilmek zorlaştı
```

```python
df.describe()   #ortalamasını alabileceği şeylerin ortalamasını aldı falan sayısal işlemler yapabildiği
```

```python
#df.mean() bunu çalıştırmaya calısırsak hata verir cünkü stringler var verilerin icinde
```

```python
df["Salary"].mean()  #salary sütunu sayısal değerler içerdiği icin oldu
```

```python
df[["Salary","Experience"]].mean()   #yine ikisininde ortalaması alınabileceğinden birlikte aldıttırdık
```

```python
df.count() #sayma işleminde hepsini sayabilir
```

```python
df[df["Experience"]>6]    #deneyimi 6 dan büyük olanların bütün df listesini getir
```

```python
df[df["Experience"]>6].count()   #tecrübesi 6 dan büyük olan kaç kişi var sorusunun cevabını verir
```
