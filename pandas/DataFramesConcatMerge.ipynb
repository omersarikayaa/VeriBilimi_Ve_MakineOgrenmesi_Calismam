```python
#MERGE KULLANIMI
```

Simdi asagidaki iki tabloda employee ve merge tablosu verildi bunlar birbirleriyle employee olaraktan ortak ama bazılarında olan çalışanlar diğer tabloda yok düzenleri aynı değil ve harici olarktan diğer name depertmant salary gibi bilgileri farklı biz bunları ortak 
olanlara göre eşitlemek isteyebiliriz merge aslında bu işe yarar

```python
df_merge = pd.read_csv("7-merge_data1.csv")
```

```python
df_merge
```

```python
df_merge2 = pd.read_csv("7-merge_data2.csv")
```

```python
df_merge2
```

```python
#merge - outer join
```

```python
df_merge_outer = pd.merge(df_merge, df_merge2, on="Employee_ID", how= "outer")
```

üstteki yaptığımız işlemde neyle neyi birleştireceğimizi verdik sonrasında on kısmınada hangi sütuna göre birleştireceğimizi verdik
outerda ise her şeyi dahil etti olmayan şeyleri NaN yaptı haricen ama tüm bilgileri merge edip dahil etti

```python
df_merge_outer
```

```python
#merge - left join
```

```python
df_merge_left = pd.merge(df_merge, df_merge2, on="Employee_ID", how="left")
```

left joinde ise ilk verdiğimiz merge'e göre eşitleme yap 10 11 12 mesela yoktu 1.tabloda ondan 2.tabloda olsada onu dahil etmeyip attı

```python
df_merge_left
```

```python
df_merge_right = pd.merge(df_merge, df_merge2, on="Employee_ID", how="right")
```

right joinde  ise 2.girdiğimiz tablodaki olmayanları attı 1 in tamtersi işte

```python
df_merge_right
```

```python
df_merge_inner = pd.merge(df_merge, df_merge2, on="Employee_ID", how="inner")
```

outer joinin tam tersi burada da her şeyi kapsamıyor direkt NaN olanları atıyor
