```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
#box plot
```

```python
# Boxplot: verinin dağılımını, medyanını, çeyreklerini ve aykırı değerlerini hızlıca gösterir.
# Median (50%): verinin ortanca değeridir.
# Q1 (25%) ve Q3 (75%): verinin alt ve üst çeyrek sınırlarıdır.
# IQR = Q3 - Q1: verinin orta %50'lik kısmının yayılımını gösterir.
# Alt sınır = Q1 - 1.5 * IQR, üst sınır = Q3 + 1.5 * IQR.
# Bu sınırların dışında kalan değerler outlier (aykırı değer) olarak kabul edilir.
# Whisker çizgileri, aykırı olmayan minimum ve maksimum değerlere kadar uzanır.
```

```python
data = np.array([5, 7, 9, 15, 20, 22, 25, 30, 32, 35, 37, 40, 50, 55, 60, 100])


plt.figure(figsize=(6,5))
# figure(figsize=(6,5)): grafiğin genişlik ve yükseklik boyutunu belirler.


sns.boxplot(y=data)
#data içerisindeki sayıların boxplot’unu oluşturuyor. y=data dediğin için dikey bir boxplot çıkıyor.

# boxplot(): verinin medyanını, çeyreklerini, yayılımını ve aykırı değerlerini görselleştirir.



plt.title("Box Plot")
plt.ylabel("Data Value")
plt.show()


# BOXPLOT ÖZETİ
# Boxplot, verinin dağılımını ve aykırı değerleri hızlıca görmemizi sağlar.

# Q1 (%25): Verilerin %25'inin altında kaldığı değerdir. Kutunun alt kenarıdır.
# Median (%50): Verinin ortanca değeridir. Kutunun içindeki çizgidir.
# Q3 (%75): Verilerin %75'inin altında kaldığı değerdir. Kutunun üst kenarıdır.

# IQR = Q3 - Q1
# IQR, verilerin ortadaki %50'lik bölümünün ne kadar yayıldığını gösterir.

# Alt sınır = Q1 - 1.5 * IQR
# Üst sınır = Q3 + 1.5 * IQR
# Bu sınırların dışında kalan değerler outlier (aykırı değer) kabul edilir.

# Whisker: Kutudan aşağı ve yukarı uzanan çizgilerdir.
# Alt whisker, aykırı olmayan en küçük değere kadar uzanır.
# Üst whisker, aykırı olmayan en büyük değere kadar uzanır.

# Bu örnekte:
# Median yaklaşık 31'dir.
# Q1 yaklaşık 18.75, Q3 yaklaşık 42.5'tir.
# Üst whisker 60'a kadar gider.
# 100 değeri üst sınırı geçtiği için outlier olarak ayrı bir nokta şeklinde gösterilir.



#100    ○      ← outlier

#60    ───    ← üst whisker
#       |
#42.5 ┌───┐  ← kutunun üstü (Q3)
#31   ├───┤  ← medyan
#18.75└───┘  ← kutunun altı (Q1)
#       |
#5     ───    ← alt whisker

```

```python
data = np.array([5, 7, 9, 15, 20, 22, 25, 30, 32, 35, 37, 40, 50, 55, 60, 100,2000])


plt.figure(figsize=(6,5))

sns.boxplot(y=data)

plt.title("Box Plot")
plt.ylabel("Data Value")
plt.show()



#outlier grafiği kötüleştirdi bunları mesela datadan çıkarmak isteyebiliriz
```
