```python
def salary_category(salary):
    if salary < 50000:
        return "low"
    elif 50000 <= salary < 80000:
        return "medium"
    else:
        return "high"
```

```python
df["Salary"].apply(salary_category)    #benim df imi Salary isine bakaraktan salary_category fonksiyonuna göre sınıflandır
```

```python
df["Salary_Category"] = df["Salary"].apply(salary_category)   #burada df tablosuna şu isimle ekle yeni oluştuduğumuz sınıflandırmayı dedik
```

şimdi hoca benden şunu isityor eğerki çalışanın experience'ı 10 dan fazla ise performance_score'u 1 arttırıp tabloya ekleyeceğim

```python
def performanceScore(ex):
    if ex>10:
        return 1
    else:
        return 0
```

```python
df["Experience"].apply(performanceScore)
```

```python
df["New_Score"] = df["Experience"].apply(performanceScore)
```

```python
df
```

```python
df["New_Score"] = df["Performance_Score"] + df["New_Score"]
```

```python
df
```

```python
def adjust_new(row):
    if row["Experience"] > 10:
        return row["Performance_Score"] +1
    else:
        return row["Performance_Score"]
```

```python
df["Adjusted_Score"] = df.apply(adjust_new , axis = 1)
```
