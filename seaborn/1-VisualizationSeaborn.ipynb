```python
plt.scatter("Height", "Weight", data=data)
#plt.scatter(data["Height"], data["Weight"]) Normalde böyle yapabilirdik.

#ama data=data dediğimiz için sütun isimlerini doğrudan verebiliyorum.

# scatter(): iki sayısal değişken arasındaki ilişkiyi noktalar halinde gösterir

# X = Height, Y = Weight


plt.xlabel("Height") # xlabel(): X ekseninin başlığını belirler

plt.ylabel("Weight") # ylabel(): Y ekseninin başlığını belirler

plt.title("Athelete Weight vs Height") # title(): grafiğe başlık ekler

plt.show() #show(): oluşturulan grafiği ekranda gösterir
```

```python
sns.set_style("whitegrid")
#Seaborn grafiğinin görünüm stilini ayarlar.
# set_style("whitegrid"): grafiğin arka planını beyaz ve ızgaralı yapar


sns.scatterplot(x="Height", y="Weight", data=data)
# scatterplot(): iki sayısal değişken arasındaki ilişkiyi noktalar halinde gösterir
# Burada X = Height, Y = Weight


plt.xlabel("Height of Atheletes")
plt.ylabel("Weight of Atheletes")
plt.title("Athletes Weight vs Height")
plt.show()
```

```python
sns.set_style("whitegrid")

sns.scatterplot(x="Height", y="Weight",hue="Sex", data=data)
#Boy ve kilo ilişkisini çiz ama noktaları Sex sütununa göre farklı renklerle ayır.
# hue: verileri belirtilen kategorik sütuna göre farklı renklerle ayırır
# Burada sporcular Sex sütununa göre M ve F olarak farklı renklerde gösterilir 
# M ve F olarak göstermesinin nedeni benim sütunumda datamada adlarının öyle olması ona göre sex dağılımını gösterdi


plt.xlabel("Height of Atheletes")
plt.ylabel("Weight of Atheletes")
plt.title("Athletes Weight vs Height")
plt.show()
```

```python
sns.set_style("whitegrid")


sns.scatterplot(x="Height", y="Weight", hue="Sex", style="Medal", data=data)
#style da hue gibi grafiğe ekstra bir değişken katıyor ama renk yerine noktanın şeklini değiştiriyor.
# hue: kategorileri farklı renklerle gösterir
# style: kategorileri farklı nokta şekilleriyle gösterir
# Burada renk = Sex, nokta şekli = Medal


plt.xlabel("Height of Atheletes")
plt.ylabel("Weight of Atheletes")
plt.title("Height vs Weight of Atheletes")
plt.show()
```

```python
sns.set_style("whitegrid")


sns.scatterplot(x="Height", y="Weight", hue="Sex", style="Medal",size="Age", data=data)
# size: sayısal bir değişkene göre noktaların büyüklüğünü değiştirir. geri kalan üstteki ile aynı 

# Scatter plot:
# x = Height, y = Weight
# hue = Sex -> cinsiyete göre renk
# style = Medal -> madalyaya göre nokta şekli
# size = Age -> yaşa göre nokta büyüklüğü



plt.xlabel("Height of Atheletes")
plt.ylabel("Weight of Atheletes")
plt.title("Height vs Weight of Atheletes")
plt.show()
```

```python
sns.set_style("whitegrid")



sns.lineplot(x="Height", y="Weight", hue="Sex", data=data)
# lineplot(): iki değişken arasındaki değişimi/eğilimi çizgi ile gösterir
# X = Height, Y = Weight
# hue = Sex -> cinsiyetleri farklı renklerle gösterir


# scatterplot -> tek tek veri noktalarını gösterir
# lineplot    -> verideki genel değişimi/eğilimi çizgi ile gösterir



plt.xlabel("Height of Atheletes")
plt.ylabel("Weight of Atheletes")
plt.title("Height vs Weight of Atheletes")
plt.show()


# Çizginin etrafındaki açık alan, ortalama tahmininin belirsizlik/güven aralığını gösterir
```

```python
sns.set_style("white")
# set_style("white"): grafiğin arka planını sade beyaz yapar


sns.displot(x="Height", hue="Sex", data=data)
# displot(): bir değişkenin dağılımını/frekansını gösterir
# X = Height, Y = bu boy aralığındaki gözlem sayısı (Frequency)
# hue="Sex": erkek ve kadın sporcuların dağılımını farklı renklerle gösterir


plt.ylabel("Frequency")
plt.title("Athletes Heigh Distrubition")
plt.show()
```

```python
# scatterplot(): iki sayısal değişken arasındaki ilişkiyi noktalar halinde gösterir.
# lineplot(): değişkenler arasındaki genel değişimi/eğilimi çizgi halinde gösterir.
# displot(): bir değişkenin dağılımını ve hangi değerlerde yoğunlaştığını gösterir.
# hue: kategorileri farklı renklerle ayırır.
# style: kategorileri farklı nokta şekilleriyle ayırır.
# size: sayısal bir değişkene göre noktaların büyüklüğünü değiştirir.
# set_style(): grafiğin genel görünüm stilini değiştirir.
```

```python
sns.set_style("white")
sns.displot(x="Height", data=data)
plt.ylabel("Frequency")
plt.title("Athletes Heigh Distrubition")
plt.show()

#NORMAL DAĞILIM VERİYOR ÇÜNKÜ BOY NORMAL DAĞILIMA UYAR
```

```python
sns.set_style("white")

sns.displot(x="Height", hue="Sex", kind="kde", data=data)
#kde = Kernel Density Estimate. Histogramdaki sütunlar yerine verinin dağılımını yumuşak bir eğriyle gösteriyor.

plt.ylabel("Frequency")
plt.title("Athletes Heigh Distrubition")
plt.show()
```

```python
sns.barplot(x="Medal", y="Height" , hue="Sex", data=data)
# barplot(): kategorilere göre sayısal değerlerin ortalamasını sütunlarla gösterir;
#burada Medal kategorilerine göre ortalama Height gösterilir ve hue="Sex" cinsiyetleri renklerle ayırır.


plt.title("Medals by Height")
plt.show()
```

```python
sns.catplot(x="Medal", y="Height", hue="Sex", col="Season", data=data)
# catplot(): kategorik verileri karşılaştırır; hue renklerle ayırır, col ise belirtilen kategoriye göre ayrı grafikler oluşturur.

# Medal kategorilerine göre Height değerlerini gösterir; hue="Sex" cinsiyete göre renklendirir
# col="Season" Summer ve Winter için ayrı grafik oluşturur.

# catplot() varsayılan olarak kind="strip" kullanır ve gözlemleri noktalar halinde gösterir.
```

```python
sns.catplot(x="Medal", y="Height", hue="Sex", col="Season",kind="bar", data=data)

# kind="bar": kategorilere göre sayısal değerlerin ortalamasını sütunlarla gösterir.
```
