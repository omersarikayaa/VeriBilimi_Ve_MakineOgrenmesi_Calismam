```python
#z-score
```

```python
data = np.array([10,12,13,15,18,25,100,105])
```

```python
#outlier(çıkıntı)
```

```python
mean=np.mean(data)  #ortalama alıyor
```

```python
mean
```

```python
data.sum()/len(data)   #manuel ortalama alıyor
```

```python
std = np.std(data)
```

```python
std
```

```python
z_score = (data-mean) / std
```

```python
z_score
```

```python
z_score > 1
```

```python
z_score[z_score > 1]
```

```python
data[z_score > 1]
```
