```python
my_list1=[1,2]
```

```python
my_list2=[2,3]
```

```python
my_list1+my_list2    #normal python dizilerde + işlemini yaptığımızda toplama yapmıyor ardına ekliyor
```

```python
my_numpy_array1 = np.array(my_list1)
```

```python
my_numpy_array2 = np.array(my_list2)
```

```python
my_numpy_array1 + my_numpy_array2   #görüğümüz üzere burada numpy arrayine çevirdiğimizden aynı indexteki elemanları topladı
```

```python
my_numpy_array1 * my_numpy_array2   #görüğümüz üzere burada numpy arrayine çevirdiğimizden aynı indexteki elemanları çarptı
```

```python
my_numpy_array1 - my_numpy_array2   #görüğümüz üzere burada numpy arrayine çevirdiğimizden aynı indexteki elemanları çıkarttı
```

```python
my_numpy_array1 / my_numpy_array2   #görüğümüz üzere burada numpy arrayine çevirdiğimizden aynı indexteki elemanları böldü
```
