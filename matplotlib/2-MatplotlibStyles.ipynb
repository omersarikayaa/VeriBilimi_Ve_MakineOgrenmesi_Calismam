```python
import numpy as np
```

```python
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
data1 = np.linspace(0,10,20)   #0 la 10 arasında 20 tane eşit aralıklarla bölünmüş sayı veriyor
```

```python
data1
```

```python
data2 = data1**2
```

```python
plt.subplots()
```

```python
my_fig, my_axes = plt.subplots()
```

```python
my_fig, my_axes = plt.subplots()   #burada tuvali oluşturuyoruz my_fig → bütün Figure, yani kağıt/tuval my_axes → grafiği çizeceğimiz alan

my_axes.plot(data1 , data2, color="#170404", alpha=0.5)    #önce data1 i x eksenine data2 yi y eksenine yazdırdık sonra altta da tam tersi
                                                           #color ile rengi verdik alpha ile de saydamlaştırabiliyoruz  

my_axes.plot(data2 , data1)
plt.show()                         #bu da işte görmemizi sağlıyor
```

```python
new_figure , new_axes = plt.subplots()

new_axes.plot(data1, data1 + 2 , color="blue", linewidth=1.0, linestyle="--", marker="o", markersize=8, markerfacecolor="yellow")
#BURADA ÖNCE X VE Y DEĞERLERİLİNİ VERDİK GRAFİK BLUE OLSUN ÇİZGİ KALINLIĞI 1.0 OLSUN ÇİZGİNİN GİDİŞİ ------- OLSUN DATAPOINTLER'İ
#O OLARAK GÖSTERİLSİN BUNLARIN BÜYÜKLÜĞÜ 8 OLSUN VE RENKLERİ SARI OLSUN DEDİK

new_axes.plot(data1, data1 + 3 , color="red" , linewidth=4.0)

plt.show()
```
