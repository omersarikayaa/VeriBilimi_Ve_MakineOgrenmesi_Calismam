```python
numpy_arr1 = np.linspace(0,10,20)  #0 dan 10 a kadar yaz 20 eş parçaya böl bu aralıktakileri ve yazdır
```

```python
numpy_arr1
```

```python
numpy_arr2 = numpy_arr1**3  
```

```python
numpy_arr2
```

```python
plt.plot(numpy_arr1, numpy_arr2, "-*")   #üssel bir görüntü için yaptık aşağıdaki grafik üssel bir de mesela tam sayı
                                         #yazdı x eksenine ama aslında tam sayı değil bu da grafiğin okunabilirliğiyle
                                         #alakalı ama aslında * olan değerler grafiğin datapointlerine karşılık gelıyor
plt.show()
```

```python
plt.plot(numpy_arr2, numpy_arr1)      # yerini değiştirince tabi ki x ve y kısımları değiştiği için değişti
plt.show()
```

```python
plt.subplot(1,2,1)    #böyle yazarkatan oluşturduğumuz grafikleri yanyana gösterebiliriz (1,2,1)
                      #okunuşu 1 satır 2 sütun 1.grafik diye


plt.plot(numpy_arr1, numpy_arr2, "r*-")
plt.subplot(1,2,2)
plt.plot(numpy_arr2, numpy_arr1, "y--")
plt.show()
```

```python
fig = plt.figure()

ax = fig.add_axes([0.1, 0.1, 0.3, 0.3])    #2.indexte yazan yükseklik 0.3 olan diğer 3.indexte yazansa yükseklik 0.3 
                                           #olan yine

ax.plot(numpy_arr1, numpy_arr2, "b")
ax.set_xlabel("X Axis")
ax.set_ylabel("Y Axis")
ax.set_title("Ornek")

plt.show()
```

```python
new_fig = plt.figure(dpi=100)     #figure oluşturuyoruz tuval ile özelleştirebiliyorduk parantez içindeki dpi çözünürlük demek
                                  #Figure = Üzerine grafik koyacağımız beyaz kağıt / tuval


new_ax = new_fig.add_axes([0.1,0.1, 0.9,0.9])    #Biraz önce new_fig ile kağıdı oluşturduk.
                                                 #Şimdi o kağıdın içine grafiğin çizileceği alanı oluşturuyoruz.



new_ax.plot(numpy_arr1 , numpy_arr1**2, label="numpy array ** 2")   #iki grafiğide ayrı ayrı çizdridik (x ekseni ,y ekseni ,label)
new_ax.plot(numpy_arr1 , numpy_arr1**3, label="numpy array ** 3")

new_ax.legend(loc=2)  #label ile ekrana yazdıklarımızın görünmesini sağladı loc=2 konumu belirledi 1 2 3 4 5 6 dedikçe yeri değişiyor

plt.show()
```
