```python
import pandas as pd  #pandası import ettik burada da 
```

```python
grades = {"Omer" : 100, "Nazif" : 50, "Efe" : 40}  #burada normal bildiğimiz sözlüğü oluşturduk
```

```python
pd.Series(grades)    #seriler tek bir sütunondan oluşmuş row coulmn şeklinde oluşmuş yapıdır
```

```python
names = ["Omer", "Nazif" , "Efe"]        #BU SEFERDE SÖZLÜK DEĞİLDE DİZİ GİBİ YAZDIM
grades = [50, 60, 30]
```

```python
pd.Series(names)                         #dizi olduğundan indexleriyle birlikte verdi
```

```python
pd.Series(grades)                        #burada da aynı şekilde yine indexleriyle verdi
```

```python
pd.Series(grades,names)       #ilk yazdığımız data ikinci yazdığımız ise index oluyor
```

```python
pd.Series(names,grades)       #burada da üstte dediğimi kanıtlıyorum
```

```python
pd.Series(data=grades, index=names)       #burada da direkt olaraktan data bu index bu olacak diye belirtebiliriz
```

```python
pd.Series(data=names, index=grades)      #üsttekinin aynısı onu kanıtlar nitelikte
```

```python
pd.Series(index=names, data=grades)       #burada da index ve datanın konumunun bu yazım tarzında önemi olmadığını..
```

ARİTMETİK

```python
contest_result = pd.Series(data = [10,20,30] , index = ["Efe" , "Nazif" , "Omer"])
```

```python
contest_result2 = pd.Series(data = [100,200,300] , index = ["Efe" , "Nazif" , "Omer"])
```

```python
contest_result
```

```python
contest_result2
```

```python
contest_result + contest_result2   #TOPLAMA ÇIKARMA ÇARPMA BÖLME TÜM ARİTMETİK İŞLEMLERİ YAPABİLİRİZ İNDEX DEĞERLERİ AYNI ZATEN
```

```python
contest_result - contest_result2
```

```python
contest_result * contest_result2
```

```python
contest_result / contest_result2
```

```python
print(contest_result["Omer"])   #böyle indexe denk gelen değeri de alabiliriz
```
